```python
#elif 
#allow u to check multiple conditions

age = 20

if age > 18:
    print(" ur an adult")
elif age > 13:
    print(" ur a teenager")
elif age > 3:
    print(" ur a kid")
elif age > 0:
    print(" ur a baby")
else:
    print(" invalid age")
```

```python
#nested if
num = int(input(" enter a number"))

if num > 0:
    print("positive number")
    if num % 2 == 0:
        print(" even number")
    else:
        print(" odd number")

else:
    print("the number is negative")
    
```

```python
##use cases 
#determine year is leap year or not

year = int(input("enter a year"))

if year %4 == 0:
    if year % 100 == 0:
        if year % 400 ==0:
            print(" leap year")
        else:
            print(" not a leap year")
    else:
        print(" leap year")
else:
    print(" not a leap year")   
    
```

```python
#simple calculator 
num = int(input("enter first number"))
num2 = int(input("enter second number"))

operation = input("Enter operation (+, -, *, /): ")

if operation == '+':
    print(" sum is", num + num2)
elif operation == '-':
    print(" difference is", num - num2)
elif operation == '*':
    print(" product is", num * num2)
elif operation == '/':
    if num2 != 0:
        print(" quotient is", num / num2)
    else:
        print(" error: division by zero")
else:
    print(" invalid operation")

```
