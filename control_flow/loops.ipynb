```python
##while loop

##while loop contidition to execute as long as the condition is true

count = 0

while count < 5:
    print("count is:", count)
    count= count + 1
```

```python
count = 0

while count%2 == 0:
    print("count is:", count)
    count= count + 1
```

```python
#loop control statements

#break statement

for i in range(10):
    if i == 5:
        break
    print(i)
```

```python
#continue statement
#continue statement skips the current iteration and moves to the next iteration of the loop
for i in range(10):
    if i % 2 == 0:
        continue
    print(i)
```

```python
##pass
#pass statement is used as a placeholder for future code. When the pass statement is executed, nothing happens, but you avoid getting an error when empty code is not allowed.
for i in range(5):
    if i==3:
        print("this is pass block",i)
        pass
    print(i)
```

```python
#nested loops
#a loop inside another loop is called nested loop

for i in range(3):
    for j in range(2):
        print("i:",i,"j:",j)
```

```python
##example :calculate the sum of first n natural numbers using a while or for loop

##while loop

n = 10

sum = 0

i = 1

while i <= n:
    sum = sum+i
    i = i + 1
print("Sum of first", n, "natural numbers is:", sum)
```

```python
##for loop

n = 10

sum = 0

for i in range(1,n+1):
    sum = sum + i
print("Sum of first", n, "natural numbers is:", sum)
```

```python
##example - prime number between a 1 to 100

for num in range(1,101):
    if num > 1:
        for i in range(2,num):
            if (num % i ) == 0:
                break
        else:
            print(num)      
```
