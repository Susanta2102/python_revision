```python
a = 10
b = 5

add_result = a + b
print("Addition:", add_result)
print("Type of Addition result:", type(add_result))

sum_result = a - b
print("Subtraction:", sum_result)
print("Type of Subtraction result:", type(sum_result))

mul_result = a * b
print("Multiplication:", mul_result)
print("Type of Multiplication result:", type(mul_result))

div_result = a / b
print("Division:", div_result)
print("Type of Division result:", type(div_result))

mod_result = a % b
print("Modulus:", mod_result)
print("Type of Modulus result:", type(mod_result))

exp_result = a ** b
print("Exponentiation:", exp_result)
print("Type of Exponentiation result:", type(exp_result))

floor_div_result = a // b
print("Floor Division:", floor_div_result)
print("Type of Floor Division result:", type(floor_div_result))
```
