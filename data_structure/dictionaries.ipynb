```python
#creating dictionaries
empty_dict = {}
print(type(empty_dict))
```

```python
student = {"name": "Alice", "age": 20, "courses": ["Math", "CompSci"]}
print(student)  
print(type(student))
```

```python
student = {"name": "Alice", "age": 20, "name": "Bob"}    #single key is repeated, last value is taken
print(student)  #{'name': 'Bob', 'age': 20} - last
print(student["name"])  #Bob
```

```python
#accessing values
student = {"name": "Alice", "age": 20, "courses": ["Math", "CompSci"]}
print(student["courses"])  #['Math', 'CompSci']
print(student.get("age"))  #20  
```

```python
#access using get() method
print(student.get("name"))  #Alice
print(student.get("phone"))  #None (key doesn't exist)
print(student.get("phone", "Not Found"))  #Not Found (default value)
    
```

```python
#modifying dictionaries elements
student["age"] = 21  #update age
student["phone"] = "555-5555"  #add new key-value pair
print(student)  
student["address"] = "123 Main St"  #add another key-value pair
print(student)
```

```python
del student["courses"]  #remove key-value pair
print(student)
```

```python
keys=student.keys()  #get all keys 
print(keys)
values=student.values()  #get all values
print(values)

items=student.items()  #get all key-value pairs
print(items)

```

```python
##shallow copy of dictionary 
#both variables point to the same dictionary object in memory

student2 = student  #both refer to same dictionary
student2["name"] = "Charlie"
print(student)  #name is changed in original dictionary as well
print(student2)
```

```python
student_copy = student.copy()  #create a shallow copy
student_copy["name"] = "David"
print(student)  #original dictionary remains unchanged
print(student_copy)  #copied dictionary has the new name
```

```python
##shawwllow copy provides a new object but nested objects are referenced
```

```python
#iterating through a dictionary
#you can iterate through keys, values, or key-value pairs
for key in student:
    print(key, student[key])  #prints key and corresponding value
for key, value in student.items():
    print(key, value)  #prints key and corresponding value
for value in student.values():
    print(value)  #prints all values
for key in student.keys():
    print(key)  #prints all keys
    
```

```python
#nested dictionaries
students = {
    "student1": {"name": "Alice", "age": 20},
    "student2": {"name": "Bob", "age": 22},
}
print(students["student1"]["name"])  #Alice
print(students["student2"]["age"])   #22        
```

```python
#itereting through nested dictionaries
for student_id, info in students.items():
    print(student_id)
    for key, value in info.items():
        print(f"  {key}: {value}")
    
    
```

```python
#dictionary comprehensions
squares = {x: x*x for x in range(6)}
print(squares)  #{0: 0, 1: 1, 2: 4, 3: 9, 4: 16, 5: 25}
print(type(squares))  #<class 'dict'>
```

```python
#conditional dictionary comprehensions
even_squares = {x: x*x for x in range(6) if x % 2 == 0}
print(even_squares)  #{0: 0, 2: 4, 4: 16}
```

```python
##practice problems
#1. Create a dictionary representing a book with keys: title, author, year, and genres (list of genres).
book = {
    "title": "The Great Gatsby",
    "author": "F. Scott Fitzgerald",
    "year": 1925,
    "genres": ["Novel", "Historical"]
}
print(book)
#2. Update the book dictionary to add a new key-value pair for publisher.
book["publisher"] = "Scribner"
print(book)
#3. Remove the year key from the book dictionary.
del book["year"]
print(book)
#4. Iterate through the book dictionary and print each key-value pair.
for key, value in book.items():
    print(f"{key}: {value}")    
```

```python
#merge two dictionaries representing two books into a single dictionary.
book1 = {
    "title": "1984",
    "author": "George Orwell",
    "year": 1949
}               
book2 = {
    "title": "To Kill a Mockingbird",
    "author": "Harper Lee",
    "year": 1960
}
merged_book = {**book1, **book2}  #book2 values will overwrite book1 values for same keys
print(merged_book)
merged_book_list = [book1, book2]  #list of book dictionaries
print(merged_book_list)
```
