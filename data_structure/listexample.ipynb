```python
#example of list
to_do_list = ["buy groceries", "clean the house", "pay bills"]
print(to_do_list)  #['buy groceries', 'clean the house', 'pay bills
print(type(to_do_list))  #<class 'list'>
```

```python
#adding tasks to the list
to_do_list.append("walk the dog")
to_do_list.append("read a book")
to_do_list.insert(1, "pay rent")  #inserting at specific position
#removing a task from the list
to_do_list.remove("pay bills")

#checking if the task is in the list
if "clean the house" in to_do_list:
    print("Task found!")
 
print( "to_do_list")  #['buy groceries', 'clean the house', 'walk the dog', 'read a book']
for task in to_do_list:
    print(task)
```
