```python
##slicing list 
numbers = [10, 20, 30, 40, 50, 60, 70, 80, 90]
print(numbers[2:5])  # Output: [30, 40, 50
print(numbers[:4])   # Output: [10, 20, 30, 40]
print(numbers[5:])   # Output: [60, 70, 80,
print(numbers[-3:])  # Output: [70, 80, 90]
print(numbers[::2])  #jump by 2  Output: [10, 30, 50, 70, 90]
print(numbers[::-1]) #reverse the list Output: [90, 80, 70
print(numbers[1::-1]) # Output: [20, 10] #printing in reverse order from index 1 to 0
print(type(numbers))
```

```python
for number in numbers:
    print(number)
```

```python
for index in range(len(numbers)):
    print("Index:", index, "Value:", numbers[index])
```

```python
for index,number in enumerate(numbers):
    print("Index:", index, "Value:", number)
```
