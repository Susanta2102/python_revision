```python
my_set = {1, 2, 3, 4, 5}
print(my_set)
print(type(my_set))
```

```python
my_empty_set = set([1, 2, 3])
print(my_empty_set)
print(type(my_empty_set))
```

```python
my_set = {1,1,1,2,2,3,4,5}
print(my_set)
print(type(my_set))  #it avooids duplicate values
```

```python
##adding elements to set
my_set.add(6)
print(my_set)
```

```python
##removing elements from set
my_set.remove(3)
print(my_set)  
```

```python
my_set.discard(10)  #does not raise error if element not found
print(my_set)
```

```python
#pop method 
popped_element = my_set.pop()
print("Popped element:", popped_element)
print("Set after pop:", my_set) 
```

```python
#clear all elements from set
my_set.clear()
print("Set after clear:", my_set)
```

```python
#set membership  #in my_set
my_set = {1, 2, 3, 4, 5}
print(3 in my_set)  #True
print(10 in my_set)  #False
```

```python
set1 = {1, 2, 3}
set2 = {3, 4, 5}
#union
union_set = set1.union(set2)
print("Union:", union_set)
#intersection
intersection_set = set1.intersection(set2)
print("Intersection:", intersection_set)
#difference
difference_set = set1.difference(set2)
print("Difference (set1 - set2):", difference_set)
#symmetric difference
symmetric_difference_set = set1.symmetric_difference(set2)
print("Symmetric Difference:", symmetric_difference_set)    
```

```python
#interation update
set1 = {1, 2, 3}
set2 = {3, 4, 5}
set1.intersection_update(set2)
print("Set1 after intersection update with Set2:", set1)
```

```python
#set operations
setA = {1, 2, 3, 4}
setB = {3, 4, 5, 6}

#is subset
print(setA.issubset(setB))  #False

#is superset
print(setA.issuperset(setB))  
```

```python
lst = [1, 2, 2, 3, 4, 4, 5]
set(lst) 
#count unique elements
unique_count = len(set(lst))
print("Number of unique elements:", unique_count)
```

```python
#counting uquique words in a sentence
sentence = "This is a sample sentence with several words this is a test"
words = sentence.lower().split()   #lower().split() : convert to lowercase and split into words
unique_words = set(words)  # Convert list of words to a set to get unique words
print("Unique words:", unique_words)
print("Number of unique words:", len(unique_words))
```
