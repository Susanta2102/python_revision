```python
#creating a tuple
my_tuple = (1, 2, 3, "hello", True)
print(my_tuple)  #(1, 2, 3, 'hello', True)
print(type(my_tuple))  #<class 'tuple'>
```

```python
my = ()
print(type(my))
```

```python
lst = [1,2,3]
print(type(lst))
tup = tuple(lst)
print(type(tup))
print(tup)
```

```python
numbers = tuple([1, 2, 3, 4, 5])
print(numbers)  #(1, 2, 3, 4, 5)
print(type(numbers))  #<class 'tuple'>
```

```python
#accessing tuple elements
my_tuple = (10, 20, 30, 40, 50)
print(my_tuple[0])  #10
print(my_tuple[2])  #30
print(my_tuple[-1])  #50
print(my_tuple[1:4])  #(20, 30, 40)
print(len(my_tuple))  #5
```

```python
#tuple operations
tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)
#concatenation
tuple3 = tuple1 + tuple2
print(tuple3)  #(1, 2, 3, 4, 5, 6)
#repetition
tuple4 = tuple1 * 2
print(tuple4)  #(1, 2, 3, 1, 2, 3)
#membership
print(2 in tuple1)  #True
print(5 not in tuple1)  #True       
```

```python
#immutability of tuples
my_tuple = (1, 2, 3)
print(my_tuple)  #(1, 2, 3)

my_tuple[0] = 10  #This will raise a TypeError
print(my_tuple)  #(1, 2, 3)
```

```python
#tuple methods
numbers = (5, 2, 3, 2, 4, 2)
#count
count_2 = numbers.count(2)
print(count_2)  #3
#index
index_3 = numbers.index(3)
print(index_3)  #2
```

```python
##packing and unpacking
#packing
person = ("Alice", 30, "Engineer")
#unpacking
name, age, profession = person
print(name)        #Alice
print(age)         #30
print(profession)  #Engineer
```

```python
#unpacking with *
data = (1, 2, 3, 4, 5)   #unpacking with rest operator
a, b, *rest = data
print(a)     #1
print(b)     #2
print(rest)  #[3, 4, 5]
```

```python
#nested tuples
nested_tuple = (1, (2, 3), (4, (5, 6)))
print(nested_tuple[1])        #(2, 3)
print(nested_tuple[1][0])     #2
print(nested_tuple[2][1][1])  #6
```

```python
#iterating through a tuple
my_tuple = (10, 20, 30, 40, 50)
for item in my_tuple:
    print(item)
#10
```
