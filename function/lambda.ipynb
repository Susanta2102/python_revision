```python
def add(x, y):
    return x + y
add(2, 3)
add_lambda = lambda x, y: x + y
add_lambda(2, 3)
print(add_lambda(2, 3))
```

```python
lambda_add = lambda x, y: x + y
result = lambda_add(5, 3)
print(f"Lambda Addition Result: {result}")
```

```python
even_numbers = list(filter(lambda x: x % 2 == 0, range(10)))
print(f"Even Numbers: {even_numbers}")
```

```python
#what is lambda function in python?
#A lambda function in Python is a small anonymous function defined using 
# the lambda keyword. It can take any number of arguments but can only have a single expression.
#  Lambda functions are often used for short
#operations where a full function definition is not necessary.
```

```python
#lambda function when we need it?
#Lambda functions are useful in scenarios where a small, throwaway function is needed for a short
# period of time. Common use cases include:
#1. As arguments to higher-order functions like map(), filter(), and reduce().
#2. For simple operations in sorting or ordering data.
#3. When defining small functions inline without cluttering the code with full function definitions.    
#4. In GUI programming for event handling where a quick function is needed.
#5. In data analysis and manipulation tasks where concise operations are required.  
```

```python
#map function

def square(x):
    return x * x
numbers = [1, 2, 3, 4, 5]
squared_numbers = list(map(square, numbers))
print(f"Squared Numbers using map and function: {squared_numbers}")
```
