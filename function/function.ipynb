```python
def function_name(parameter1, parameter2):
    """
    This function takes two parameters and returns their sum.
    """
    return parameter1 + parameter2
# Example usage
result = function_name(5, 10)
print("The sum is:", result)
```

```python
#why function are used?
#1. Code Reusability: Functions allow you to write a piece of code once and use it multiple times throughout your program, reducing redundancy.
#2. Modularity: Functions help break down complex problems into smaller, manageable parts, making the code easier to understand and maintain.
#3. Readability: By using functions, you can give meaningful names to blocks of code, which enhances the readability of your program.
#4. Abstraction: Functions allow you to hide the implementation details and expose only the necessary interface, making it easier to work with complex systems.
#5. Testing and Debugging: Functions can be tested individually, making it easier to identify and fix bugs in specific parts of the code.
#6. Organization: Functions help organize code into logical sections, making it easier to navigate and manage larger codebases. 
```

```python
def even_or_odd(num):
    if num % 2 == 0:
        return "Even"
    else:
        return "Odd"
# Example usage
number = 7
result = even_or_odd(number)
print(f"The number {number} is {result}.")
```

```python
#variable length arguments
#positional and keyword arguments

def print_numbers(*args):
    for number in args:
        print(number)
    
print_numbers(1, 2, 3, 4, 5, "susant")
```

```python
#keywords arguments
def print_info(name, age):
    print("Name:", name)
    print("Age:", age)
print_info(age=25, name="John") 
# Output:
```

```python
def print_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")
print_details(name="Alice", age=30, city="New York")
```

```python
def print_info(*args, **kwargs):
    for arg in args:
        print("Positional argument:", arg)
    for key, value in kwargs.items():
        print(f"Keyword argument - {key}: {value}")

```

```python
#difference between key argument and positional argument

# Positional arguments: Passed to the function in the exact order the parameters are defined.
# They are matched by position, not by name.
# Example:
def add(a, b):
    return a + b

result1 = add(5, 10)  # a=5, b=10
print("Positional:", result1)

# Keyword arguments: Passed by explicitly naming the parameter, allowing any order.
# They are matched by name, not by position.
# Example:
result2 = add(b=20, a=15)  # a=15, b=20
print("Keyword:", result2)

# You can mix them, but positional must come before keyword arguments.
result3 = add(25, b=30)  # a=25, b=30
print("Mixed:", result3)
```
