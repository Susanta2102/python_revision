```python
def square(x):
    return x * x
numbers = [1, 2, 3, 4, 5]
squared_numbers = list(map(square, numbers))
print(f"Squared Numbers using map and function: {squared_numbers}")
```

```python
#labda function with map function

squared_numbers_lambda = list(map(lambda x: x * x, numbers))
print(f"Squared Numbers using map and lambda: {squared_numbers_lambda}")
```

```python
##map multiple iterable
numbers1 = [1, 2, 3]
numbers2 = [4, 5, 6]
numbers3 = [7, 8, 9]
summed_numbers = list(map(lambda x, y, z: x + y + z, numbers1, numbers2, numbers3))
print(f"Summed Numbers from three lists: {summed_numbers}")
```

```python
#map to convert list of strings to integers
string_numbers = ["1", "2", "3", "4", "5"]
int_numbers = list(map(int, string_numbers))
print(f"Converted Integers: {int_numbers}")
print(type(int_numbers[0]))  # Output: <class 'int'>
```

```python
words = ["hello", "world", "python"]
uppercased_words = list(map(lambda word: word.upper(), words))
print(f"Uppercased Words: {uppercased_words}")
```

```python
def get_name_length(name):
    return len(name)
names = ["Alice", "Bob", "Charlie"]
name_lengths = list(map(get_name_length, names))
print(f"Name Lengths using map and function: {name_lengths}")   
```
